```python
# Binary Tree
class Node:
  def __init__(self,data):
    self.data=data
    self.left=None
    self.right= None
```

```python
# left,print root,right
def inorder(node):
  if node is None:
    return

  inorder(node.left)
  print(node.data)
  inorder(node.right)
```

```python
# root print,left,right
def preorder(node):
  if node is None:
    return

  print(node.data)
  preorder(node.left)
  preorder(node.right)
```

```python
# left,right,root print
def postorder(node):
  if node is None:
    return

  postorder(node.left)
  postorder(node.right)
  print(node.data)

```

```python
# find max dist. b/w two nodes (links not nodes)
# find diameter

max_len=0
def find_distt(root):
  global max_len
  if not root:
    return 0
  max_left= find_distt(root.left)
  max_right= find_distt(root.right)
  max_len = max(max_len,max_left+max_right)
  return 1+max(max_left,max_right)

print(find_distt(root))
print(max_len)
```

```python
# find height
def find_height(root):
  if not root:
    return 0
  max_left= find_height(root.left)
  max_right= find_height(root.right)
  return 1+max(max_left,max_right)
```

```python
# findmax in tree
def find_max(node):
  if node is None:
    return -1000

  leftree= find_max(node.left)
  rightree= find_max(node.right)

  return max(leftree,node.data,rightree)
```
